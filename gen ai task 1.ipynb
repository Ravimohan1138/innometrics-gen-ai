```python
# ASSIGNMENT: Prompt Templates using LangChain

!pip install -q langchain


from langchain_core.prompts import PromptTemplate
from langchain_core.prompts import ChatPromptTemplate


# TASK 1: Replace Hardcoded Prompts

print("========== TASK 1 ==========\n")

template_1 = PromptTemplate(
    input_variables=["topic"],
    template="Explain {topic} in simple terms for beginners"
)

result_1 = template_1.format(
    topic="Artificial Intelligence"
)

print(result_1)


# TASK 2: Multi-Input Prompt System

print("\n========== TASK 2 ==========\n")

template_2 = PromptTemplate(
    input_variables=["topic", "audience", "tone"],
    template="Explain {topic} for {audience} in a {tone} tone"
)

# Test Case 1
output_1 = template_2.format(
    topic="AI",
    audience="beginners",
    tone="friendly"
)

print("Test Case 1:")
print(output_1)
print()


# Test Case 2
output_2 = template_2.format(
    topic="Python",
    audience="kids",
    tone="fun"
)

print("Test Case 2:")
print(output_2)
print()


# Test Case 3
output_3 = template_2.format(
    topic="Deep Learning",
    audience="engineers",
    tone="technical"
)

print("Test Case 3:")
print(output_3)


# TASK 3: Prompt Variations Engine

print("\n========== TASK 3 ==========\n")

teaching_template = PromptTemplate(
    input_variables=["topic"],
    template="Explain {topic} clearly step by step"
)

interview_template = PromptTemplate(
    input_variables=["topic"],
    template="Ask 3 interview questions about {topic}"
)

story_template = PromptTemplate(
    input_variables=["topic"],
    template="Explain {topic} as a story"
)

topic_name = "Machine Learning"

print("Teaching Prompt:")
print(teaching_template.format(topic=topic_name))
print()

print("Interview Prompt:")
print(interview_template.format(topic=topic_name))
print()

print("Storytelling Prompt:")
print(story_template.format(topic=topic_name))


# TASK 4: ChatPromptTemplate System

print("\n========== TASK 4 ==========\n")

role = "teacher"
topic = "Neural Networks"

chat_template = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful {role}."),
    ("user", "Teach me about {topic}.")
])

chat_messages = chat_template.format_messages(
    role=role,
    topic=topic
)

print(chat_messages)


# TASK 5: Input Validation Layer

print("\n========== TASK 5 ==========\n")

def validate_inputs(audience, tone):

    valid_audience = ["beginner", "intermediate", "expert"]
    valid_tone = ["formal", "casual", "fun"]

    if audience not in valid_audience:
        raise ValueError(
            "Invalid audience! Choose from beginner, intermediate, expert"
        )

    if tone not in valid_tone:
        raise ValueError(
            "Invalid tone! Choose from formal, casual, fun"
        )

    return True


# Testing Validation
try:
    validate_inputs("beginner", "fun")
    print("Inputs are valid")

except ValueError as error:
    print(error)


# TASK 6: Prompt Generator App

print("\n========== TASK 6 ==========\n")

def generate_prompt(topic, audience, tone, style):

    templates = {

        "teaching": PromptTemplate(
            input_variables=["topic", "audience", "tone"],
            template=(
                "Explain {topic} for {audience} "
                "in a {tone} teaching style"
            )
        ),

        "interview": PromptTemplate(
            input_variables=["topic", "audience", "tone"],
            template=(
                "Ask 3 interview questions about {topic} "
                "for {audience} in a {tone} style"
            )
        ),

        "storytelling": PromptTemplate(
            input_variables=["topic", "audience", "tone"],
            template=(
                "Explain {topic} for {audience} "
                "in a {tone} storytelling style"
            )
        )
    }

    selected_template = templates[style]

    final_prompt = selected_template.format(
        topic=topic,
        audience=audience,
        tone=tone
    )

    return final_prompt


generated_output = generate_prompt(
    topic="Neural Networks",
    audience="beginners",
    tone="fun",
    style="storytelling"
)

print(generated_output)


# TASK 7: Template Reusability Test


print("\n========== TASK 7 ==========\n")

reuse_template = PromptTemplate(
    input_variables=["topic"],
    template="Explain {topic} in simple terms"
)

topics = [
    "Artificial Intelligence",
    "Python",
    "Cloud Computing",
    "Machine Learning",
    "Cyber Security"
]

for item in topics:
    print(reuse_template.format(topic=item))


# FINAL CONCLUSION


print("\n========== CONCLUSION ==========\n")

print(
    "This assignment helped in understanding reusable "
    "prompt systems using LangChain PromptTemplate "
    "and ChatPromptTemplate."
)
```
